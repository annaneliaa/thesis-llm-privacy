```python
import xml.etree.ElementTree as ET
import os
import glob
```

```python
def extract_sentences_from_xml(file_path):
    # Define the namespace
    ns = {'tei': 'http://www.tei-c.org/ns/1.0'}

    # Parse the XML content from the file
    tree = ET.parse(file_path)
    root = tree.getroot()

    # Find all <seg> elements with type="original" and extract their text
    sentences = []
    for seg in root.findall('.//tei:seg[@type="original"]', ns):
        if seg.text:
            sentences.append(seg.text.strip())

    return sentences

def process_files(directory, spanish_output_file, greek_output_file):
    # Find all files ending with -es-tei.xml or -el-tei.xml
    es_files = glob.glob(os.path.join(directory, '*-es-tei.xml'))
    el_files = glob.glob(os.path.join(directory, '*-el-tei.xml'))
    
    # Match files based on IDs and process pairs
    for es_file in es_files:
        # Extract ID from the Spanish file name
        es_id = os.path.basename(es_file)[:13]  # Extract the common identifier from the Spanish filename
        for el_file in el_files:
            # Extract ID from the Greek file name
            el_id = os.path.basename(el_file)[:13]  # Extract the common identifier from the Greek filename
            # Check if IDs match
            if es_id == el_id:
                # Process Spanish files and append sentences to the output file
                sentences = extract_sentences_from_xml(es_file)
                with open(greek_output_file, 'a', encoding='utf-8') as greek_output:
                    for sentence in sentences:
                        greek_output.write(sentence + '\n')
                # Process Greek files and append sentences to the output file
                sentences = extract_sentences_from_xml(el_file)
                with open(spanish_output_file, 'a', encoding='utf-8') as spanish_output:
                    for sentence in sentences:
                        spanish_output.write(sentence + '\n')
    
```

```python
def clean_directory(directory):
    # Define the patterns to keep and remove
    patterns_to_keep = ['*es-tei.xml', '*el-tei.xml']
    pattern_to_remove = '*-el-es-tei.xml'

    # Create a set of files to keep based on the patterns
    files_to_keep = set()
    for pattern in patterns_to_keep:
        files_to_keep.update(glob.glob(os.path.join(directory, pattern)))

    # List all files in the directory
    all_files = glob.glob(os.path.join(directory, '*'))

    # Determine files to remove
    files_to_remove = set(glob.glob(os.path.join(directory, pattern_to_remove)))
    files_to_remove.update(set(all_files) - files_to_keep)

    # Remove files that do not match the patterns to keep or match the pattern to remove
    for file_path in files_to_remove:
        os.remove(file_path)
        print(f'Removed: {file_path}')
```
